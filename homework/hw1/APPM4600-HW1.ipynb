## Problem 1

```python
#part i

#define p(x)
def p1(x):
    y = x**9 - 18*x**8 + 144*x**7 - 672*x**6 + 2016*x**5 - 4032*x**4 + 5376*x**3 - 4608*x**2 + 2304*x - 512
    return y

#define x array
x = np.arange(1.920, 2.080, 0.001)

#plot
plt.plot(x, p1(x))
plt.savefig("appm4600-hw1-fig1.png", dpi=300, bbox_inches="tight")
```

```python
#part ii

#define factored form
def p2(x):
    return (x-2)**9

#plot
plt.plot(x, p2(x))
plt.savefig("appm4600-hw1-fig2.png", dpi=300, bbox_inches="tight")
```

The plot of the compacted form gives a curve that is smooth, and less noisy compared the the plot of the expanded form. The discrepency comes from more rounding errors in the expanded polynomial, because you need to do more arithmetic to solve the problem in expanded form - when you subtract numbers that are close in magnitude, floating-point arithmetic loses precision. The simpler computation yields a simpler plot with more precision. 

## Problem 5

```python
#part b

delta = np.logspace(-16, 0, 200)

def og(x, d):
    return np.cos(x + d) - np.cos(x)

def new(x, d):
    return -2*np.sin(x + d/2)*np.sin(d/2)

xs = [np.pi, 1e6]
labels = [r'$x=\pi$', r'$x=10^6$']

fig, axes = plt.subplots(1, 2, figsize=(12,5))

for ax, x, label in zip(axes, xs, labels):
    og_vals = og(x, delta)
    new_vals = new(x, delta)
    diff = np.abs(og_vals - new_vals)

    ax.loglog(delta, diff, marker='o', markersize=3, linewidth=1)
    ax.set_xlabel(r'$\delta$')
    ax.set_ylabel(r'$|\,\mathrm{original} - \mathrm{new}\,|$')
    ax.set_title(f'Difference between methods, {label}')
    ax.grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.savefig("appm4600-hw1.png", dpi=300, bbox_inches="tight")
```
